`max(1, 25 − ln(params + bytes + MACs))` — smaller networks score higher.
 
| Architecture | Cost | Score |

````python
"""
NeuroGolf 2026 — Minimal Submission
Only the two datasets that proved to give 5263.65
NO validation, NO extra sources, NO complexity
"""
import math, zipfile, re, os
from pathlib import Path
from onnx import helper, TensorProto

MAX_BYTES = int(1.44 * 1024 * 1024)
BANNED    = {"Loop","Scan","NonZero","Unique"}
TASK_RE   = re.compile(r"^task(\d{3})\.onnx$")

try:
    import onnx
    import onnxruntime
except ImportError:
    os.system("pip install -q onnx onnxruntime")
    import onnx

# ── ONLY these two sources — proven to give 5263 ──────────
SOURCES = [
    Path("/kaggle/input/datasets/jonathanchan/sub4979-07"),
    Path("/kaggle/input/datasets/jonathanchan/sub4971-98"),
]

def compute_cost(raw: bytes):
    if len(raw) > MAX_BYTES: return None
    try:
        m = onnx.load_model_from_string(raw)
        m = onnx.shape_inference.infer_shapes(m)
    except: return None
    tb={1:4,6:4,7:8,10:2,11:8,16:2,2:1,3:1,4:2,5:2}
    p=nb=mac=0; sh={}; ops=set()
    for i in m.graph.initializer:
        dims=list(i.dims); n=math.prod(dims) if dims else 1
        p+=n; nb+=n*tb.get(i.data_type,4); sh[i.name]=dims
    for i in m.graph.input:
        s=[]
        for d in i.type.tensor_type.shape.dim:
            if d.HasField('dim_value'): s.append(d.dim_value)
            else: return None
        sh[i.name]=s
    for vi in list(m.graph.value_info)+list(m.graph.output):
        s=[]
        for d in vi.type.tensor_type.shape.dim:
            if d.HasField('dim_value'): s.append(d.dim_value)
            else: return None
        if s:
            n=math.prod(s); nb+=n*tb.get(vi.type.tensor_type.elem_type,4); sh[vi.name]=s
    for node in m.graph.node:
        ops.add(node.op_type)
        if node.op_type=="Conv" and len(node.input)>=2:
            o=sh.get(node.output[0],[]); w=sh.get(node.input[1],[])
            if len(o)==4 and len(w)==4:
                _,co,ho,wo=o; _,ci,kh,kw=w; mac+=co*ci*kh*kw*ho*wo
        elif node.op_type in ("Gemm","MatMul") and len(node.input)>=2:
            a=sh.get(node.input[0],[]); b=sh.get(node.input[1],[])
            if len(a)>=2 and len(b)>=2: mac+=math.prod(a[:-1])*a[-1]*b[-1]
    if BANNED & ops: return None
    return p+nb+mac

# ── Load all models from both sources ─────────────────────
best = {}   # task_name -> (raw, cost)

for root in SOURCES:
    if not root.exists():
        print(f"NOT FOUND: {root}"); continue
    loaded = 0
    for zp in sorted(root.rglob("*.zip")):
        try:
            with zipfile.ZipFile(zp) as zf:
                for entry in zf.namelist():
                    fname = Path(entry).name
                    m = TASK_RE.match(fname)
                    if not m: continue
                    raw  = zf.read(entry)
                    cost = compute_cost(raw)
                    if cost is None: continue
                    stem = f"task{m.group(1)}"
                    if stem not in best or cost < best[stem][1]:
                        best[stem] = (raw, cost)
                        loaded += 1
        except: pass
    for p in sorted(root.rglob("task*.onnx")):
        m = TASK_RE.match(p.name)
        if not m: continue
        raw  = p.read_bytes()
        cost = compute_cost(raw)
        if cost is None: continue
        stem = f"task{m.group(1)}"
        if stem not in best or cost < best[stem][1]:
            best[stem] = (raw, cost)
            loaded += 1
    print(f"Loaded from {root.name}: {len(best)} total tasks")

# ── Identity fallback for any gaps ────────────────────────
def make_identity():
    X=helper.make_tensor_value_info("input", TensorProto.FLOAT,[1,10,30,30])
    Y=helper.make_tensor_value_info("output",TensorProto.FLOAT,[1,10,30,30])
    g=helper.make_graph([helper.make_node("Identity",["input"],["output"])],"id",[X],[Y])
    return helper.make_model(g,opset_imports=[helper.make_opsetid("",17)]).SerializeToString()

for i in range(1, 401):
    name = f"task{i:03d}"
    if name not in best:
        best[name] = (make_identity(), 0)
        print(f"  identity fallback: {name}")

# ── Pack zip ───────────────────────────────────────────────
OUT = Path("/kaggle/working/submission.zip")
with zipfile.ZipFile(OUT, "w", zipfile.ZIP_DEFLATED) as zf:
    for name in sorted(best.keys()):
        num = name.replace("task","")
        zf.writestr(f"task{num}.onnx", best[name][0])

est = sum(
    (max(1.0, 25.0-math.log(max(1,c))) if c>0 else 25.0)
    for _, c in best.values()
)

print(f"\nTasks  : {len(best)} / 400")
print(f"Est    : {est:.2f}")
print(f"Size   : {OUT.stat().st_size/1024:.1f} KB")
print(f"\n✓ submission.zip ready — download from Output tab")
````
